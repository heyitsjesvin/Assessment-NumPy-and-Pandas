```python
#12

# Students whose Final_Score is greater than 75
print("Final Score greater than 75:")
print(df[df['Final_Score'] > 75])

# Students whose Attendance is below 75
print("\nAttendance below 75:")
print(df[df['Attendance'] < 75])

# Data Science students with Final_Score greater than 80
print("\nData Science students with Final Score greater than 80:")
print(df[(df['Department'] == 'Data Science') & (df['Final_Score'] > 80)])

# Sort by Final_Score from highest to lowest
print("\nSorted by Final Score:")
print(df.sort_values(by='Final_Score', ascending=False))
```

```python
#13
import numpy as np

df['Average_Technical_Score'] = df[
    ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
].mean(axis=1)

conditions = [
    df['Average_Technical_Score'] >= 90,
    df['Average_Technical_Score'] >= 75,
    df['Average_Technical_Score'] >= 60
]

choices = [
    'Excellent',
    'Good',
    'Average'
]

df['Performance'] = np.select(
    conditions,
    choices,
    default='Needs Improvement'
)

print(df)
```

```python
#14
result = df.groupby('Department')['Final_Score'].agg(
    Number_of_Students='count',
    Average_Final_Score='mean',
    Maximum_Final_Score='max',
    Minimum_Final_Score='min'
)

result = result.sort_values('Average_Final_Score', ascending=False)

print(result)
```

```python
#16
#A
import pandas as pd
import numpy as np

df = pd.read_csv('/content/drive/MyDrive/DSAB13/Assesment/student_performance.csv')

# Identify missing values
print("Missing values:")
print(df.isnull().sum())

# Numerical columns
numeric_columns = [
    'Age',
    'Attendance',
    'Python_Score',
    'SQL_Score',
    'Pandas_Score',
    'NumPy_Score',
    'Final_Score'
]

# Replace missing numerical values with mean
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
#16
#B
df['Average_Score'] = df[
    ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
].mean(axis=1)

df['Pass_Status'] = np.where(
    df['Average_Score'] >= 50,
    'Pass',
    'Fail'
)

print(df)
```

```python
#16
#C
top_5 = df.sort_values(
    by='Average_Score',
    ascending=False
).head(5)

print("Top 5 students:")
print(top_5[['Student_ID', 'Name', 'Department', 'Average_Score']])
```

```python
#16
#C2
department_average = df.groupby('Department')['Average_Score'].mean()

print("Department average score:")
print(department_average)

print("\nDepartment with highest average score:")
print(department_average.idxmax())
```

```python
#16
#c3
average_attendance = df.groupby('Department')['Attendance'].mean()

print("Average attendance by department:")
print(average_attendance)
```

```python
#c
#c4
pass_fail_count = df['Pass_Status'].value_counts()

print("Pass and Fail count:")
print(pass_fail_count)
```

```python
#16
#D
summary = df.groupby('Department').agg(
    Total_Students=('Student_ID', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Pass_Count=('Pass_Status', lambda x: (x == 'Pass').sum()),
    Fail_Count=('Pass_Status', lambda x: (x == 'Fail').sum())
).reset_index()

print(summary)
```

```python
#17
#A. Data Preparation
df = pd.read_csv('/content/drive/MyDrive/DSAB13/Assesment/sales.csv')

#Convert columns to correct data types
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')

#Handle missing values and clean categories/regions
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Profit'] = df['Profit'].fillna(df['Profit'].median())
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
df['Discount'] = df['Discount'].fillna(df['Discount'].mean())
df['Region'] = df['Region'].fillna('Unknown').str.strip().str.title()
df['Category'] = df['Category'].str.strip().str.title()

# B. Sales Analysis
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
avg_sales = df['Sales'].mean()

top_5_products = df.groupby('Product')['Sales'].sum().nlargest(5)
top_region = df.groupby('Region')['Sales'].sum().idxmax()

print(f"Total Sales: {total_sales:.2f}")
print(f"Total Profit: {total_profit:.2f}")
print(f"Average Sales: {avg_sales:.2f}")
print("Top 5 Products:\n", top_5_products)
print("Region with Highest Sales:", top_region)

#- C. Category Analysis ---
cat_summary = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Discount=('Discount', 'mean')
).sort_values(by='Total_Sales', ascending=False)

print("\nCategory Summary:\n", cat_summary)
```
